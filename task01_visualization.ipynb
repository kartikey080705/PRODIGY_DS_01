```python
import pandas as pd

data = {
    "Age": [18, 20, 22, 25, 27, 30, 32, 35, 40, 45, 50, 23, 28, 33, 38],
    "Gender": [
        "Male", "Female", "Female", "Male", "Male",
        "Female", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ]
}

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["Age"], bins=6)
plt.title("Age Distribution of Sample Population")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Individuals")
plt.show()
```

```python
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(6,4))
gender_counts.plot(kind="bar")
plt.title("Gender Distribution of Sample Population")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.show()
```

### Insights:
- The age distribution shows that most individuals fall between 20 and 40 years.
- The gender distribution is relatively balanced, with a slightly higher number of males.
